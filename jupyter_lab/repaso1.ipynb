Reto 1.1: Extremos y centro
Escribe extremos(lista) que devuelva una tupla (primero, ultimo, central). Si la longitud es par, el central es el de la posición len // 2.
assert extremos([1, 2, 3]) == (1, 3, 2)
assert extremos([10, 20, 30, 40]) == (10, 40, 30)

```python
def extremos(lista):
    return (lista[0], lista[-1], lista[len(lista)//2])

print(extremos([1,2,3]))
print(extremos([10,20,30,40]))
```

Reto 1.2: Escribe rotar(lista, k) que mueva los últimos k elementos al inicio sin modificar la lista original y usando solo slicing.
original = [1, 2, 3, 4, 5]
assert rotar(original, 2) == [4, 5, 1, 2, 3]
assert original == [1, 2, 3, 4, 5]

```python
def rotate(ex_list: list, n:int) -> list:
    last_k_elements = ex_list[len(ex_list)-n: len(ex_list)]
    return last_k_elements + ex_list[:len(ex_list)-n]

rotate([1,2,3,4,5], 2)
```

Reto 2.1: Sin duplicados, mismo orden
Escribe sin_duplicados(lista) que quite los repetidos conservando el orden de aparición. Pista: set() solo no basta, porque no garantiza el orden que necesitas. Piensa cómo combinarlo con in.
assert sin_duplicados([3, 1, 3, 2, 1, 5]) == [3, 1, 2, 5]

```python
def without_duplicates(ex:list) -> list:
    aux = list(set(ex))
    aux.sort(key=ex.index)
    return aux 

def without_duplicates2(aux_list:list) -> list:
    final_list = set()
    new_list=[]
    for i in aux_list:
        if i not in final_list:
            final_list.add(i)
            new_list.append(i)
    
    return new_list


# without_duplicates([3, 1, 3, 2, 1, 5])
without_duplicates2([3, 1, 3, 2, 1, 5])
```

Reto 2.2: Aplanar
Escribe aplanar(matriz) que convierta una lista de listas en una lista simple. Hazlo primero con bucles for y luego con una comprehension.

to_plain([[1, 2], [3], [4, 5, 6]]) == [1, 2, 3, 4, 5, 6]


```python
def to_plain(matrix:list[list]) -> list:
    plain_list = []
    for row in matrix:
        for element in row:
            plain_list.append(element)
    return plain_list

to_plain([[1, 2], [3], [4, 5, 6]])

def comprenhension_to_plain(matrix:list[list]) ->list:
    return [element for row in matrix for element in row]

to_plain([[1, 2], [3], [4, 5, 6]])
```
