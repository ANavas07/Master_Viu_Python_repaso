Crear una lista con números que comiencen por 30 y que termine en 36 haciendo que el paso entre cada número consecutivo sea de 0.4. Sin usar Numpy.

# Your Solution Here!!!

```python
def frange(initial, last_number, salt):
    response = []
    actual = float(initial)

    while actual <= last_number:
        response.append(round(actual, 10))
        actual += salt
    
    return response

print(frange(30,36,0.4))
```
